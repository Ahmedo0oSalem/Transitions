```python
import bz2
import json

frames = []

with bz2.open(path, "rt") as f:
    for i in range(5):
        frames.append(json.loads(next(f)))

for i, frame in enumerate(frames):
    print(f"\n========== Frame {i} ==========")
    for key in frame:
        print(f"{key}:")
        print(frame[key])
        print()
```
